```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/students.csv.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
average_marks = df[["Math","Science","English"]].mean()
print(average_marks)
```

```python
highest_subject = average_marks.idxmax()
```

```python
print("Highest scoring subject:", highest_subject)
```

```python
df["Average"] = df[["Math", "Science", "English"]].mean(axis=1)
```

```python
weak_students= df[df["Average"] < 70]
print(weak_students)
```

```python
average_marks.plot(kind="bar")
plt.title("average marks per subject")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.show()
```

```python
df.set_index("Name")[["Math","Science","English"]].plot(kind="bar")
plt.title("student performence")
plt.ylabel("Marks")
plt.show()
```
